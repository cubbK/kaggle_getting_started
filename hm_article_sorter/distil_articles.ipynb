```python
import pandas as pd

articles = pd.read_csv('dataset/articles.csv')
articles
```

```python
import os
from pathlib import Path

def check_image_exists(article_id):
    """
    Check if an image exists for a given article_id.
    Images are stored in dataset/images/{first_3_digits}/{article_id}.jpg
    """
    # Convert to string and pad with zeros if needed
    article_id_str = str(article_id).zfill(10)  # Ensure 10 digits
    
    # Get first 3 digits for folder name
    folder_name = article_id_str[:3]
    
    # Construct the full path
    image_path = Path(f"dataset/images/{folder_name}/{article_id_str}.jpg")
    
    return image_path.exists()

# Apply the function to check image existence for all articles
print("Checking image existence for all articles...")
articles['image_exists'] = articles['article_id'].apply(check_image_exists)

# Display summary
total_articles = len(articles)
articles_with_images = articles['image_exists'].sum()
articles_without_images = total_articles - articles_with_images

print(f"\nSummary:")
print(f"Total articles: {total_articles}")
print(f"Articles with images: {articles_with_images}")
print(f"Articles without images: {articles_without_images}")
print(f"Percentage with images: {articles_with_images/total_articles*100:.2f}%")

# Show first few rows with the new column
print(f"\nFirst 10 articles with image existence check:")
print(articles[['article_id', 'image_exists']].head(10))
```

```python
# Remove rows that don't have images
print(f"Original number of articles: {len(articles)}")

# Filter to keep only articles with images
articles_with_images_only = articles[articles['image_exists'] == True].copy()

print(f"Articles after removing those without images: {len(articles_with_images_only)}")
print(f"Removed {len(articles) - len(articles_with_images_only)} articles without images")

# Update the original dataframe
articles = articles_with_images_only

# Verify all remaining articles have images
print(f"\nVerification - All remaining articles have images: {articles['image_exists'].all()}")
print(f"Final dataset shape: {articles.shape}")

# Show first few rows
print(f"\nFirst 5 rows of cleaned dataset:")
articles[['article_id', 'prod_name', 'image_exists']].head()
```

```python
# Create a column with the image path for each article
def get_image_path(article_id):
    """
    Get the image path for a given article_id.
    Images are stored in dataset/images/{first_3_digits}/{article_id}.jpg
    """
    # Convert to string and pad with zeros if needed
    article_id_str = str(article_id).zfill(10)  # Ensure 10 digits
    
    # Get first 3 digits for folder name
    folder_name = article_id_str[:3]
    
    # Construct the full path
    image_path = f"dataset/images/{folder_name}/{article_id_str}.jpg"
    
    return image_path

# Add the image_path column
articles['image_path'] = articles['article_id'].apply(get_image_path)

print("Added image_path column!")
print(f"Dataset shape: {articles.shape}")

# Show first few rows with the new image_path column
print(f"\nFirst 5 rows with image paths:")
articles[['article_id', 'prod_name', 'image_exists', 'image_path']].head()
```

```python
# Export the filtered dataset to dataset_filtered/articles.csv
import os

# Create the dataset_filtered directory if it doesn't exist
output_dir = "dataset_filtered"
os.makedirs(output_dir, exist_ok=True)

# Export the filtered articles to CSV
output_path = os.path.join(output_dir, "articles.csv")
articles.to_csv(output_path, index=False)

print(f"Filtered dataset exported to: {output_path}")
print(f"Number of articles exported: {len(articles)}")
print(f"Number of columns: {len(articles.columns)}")
print(f"File size: {os.path.getsize(output_path) / (1024*1024):.2f} MB")

# Show the columns that were exported
print(f"\nColumns exported:")
print(list(articles.columns))
```
