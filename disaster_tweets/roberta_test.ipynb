```python
import pandas as pd

import keras
import keras_hub
import numpy as np
from sklearn.model_selection import train_test_split

# Create the pretrained classifier for binary classification
classifier = keras_hub.models.RobertaClassifier.from_preset(
    "roberta_base_en",
    num_classes=2,  # Binary classification: disaster (1) or not disaster (0)
)
# Load the weigts
classifier.load_weights('roberta_disaster_classifier.weights.h5')
# Load the test data
df = pd.read_csv('data/test.csv')
# Prepare the test data
X_test = df['text'].values
# Predict the classes
predictions = classifier.predict(X_test)
# Convert predictions to binary labels
predicted_labels = np.argmax(predictions, axis=1)
# Create a DataFrame for the results
results_df = pd.DataFrame({
    'id': df['id'],
    'target': predicted_labels
})
# Save the results to a CSV file
results_df.to_csv('submission_roberta.csv', index=False)
```
